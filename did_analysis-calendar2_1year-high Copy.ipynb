```python
# Settings
match_method = "calendar2_1year"
outcome_col = "top3_mean_consumption"

EVENT_WINDOW = (-12, 12)
REFERENCE_EVENT_TIME = -1
ALL_YLIM = (-1.4, 1.2)
HIGH_YLIM = (-1.4, 1.2)
LOW_YLIM = (-1.4, 1.2)
```

```python
# Read data
matches = pd.read_parquet(
    f"/lakehouse/default/Files/output/matching_high/{match_method}/matches"
)

month_result = pd.read_parquet(
    "/lakehouse/default/Files/month_data"
)
```

```python
# ============================================================
# HIGH PRICE PERIOD — using transformed new_event_time
# ============================================================

price_value = "high"

# 1. Prepare price-period panel
df_high = prepare_price_panel(
    month_result=month_result,
    price_value=price_value,
    outcome_col=outcome_col
)

# 2. Build matched panel
matched_high = build_matched_panel(
    matches=matches,
    month_result=df_high,
    outcome_col=outcome_col
)

# 3. Transform event_time into winter-aligned new_event_time
matched_high = transform_event_time_for_matching(
    matched_high,
    cohort_col="cohort",
    event_col="event_time",
    target_months=[1, 2, 3, 11, 12],
    pre_start=-1,
    new_event_col="new_event_time"
)

print("\n===== HIGH PRICE: TRANSFORMED EVENT TIME CHECK =====")
print(
    matched_high[
        [
            "aID", "cohort", "TIDPUNKT", "treatment",
            "event_time", "cohort_event_month",
            "target_month_flag", "new_event_time"
        ]
    ]
    .sort_values(["cohort", "aID", "event_time"])
    .head(100)
)

# 4. Keep only observations with valid new_event_time
matched_high = matched_high.dropna(subset=["new_event_time"]).copy()
matched_high["new_event_time"] = matched_high["new_event_time"].astype(int)

# 5. Recompute post and D based on new_event_time
matched_high["post"] = (matched_high["new_event_time"] >= 0).astype(int)
matched_high["D"] = matched_high["treatment"] * matched_high["post"]

print("\n===== HIGH PRICE: DATA USED IN DID =====")
print(
    matched_high[
        [
            "aID", "cohort", "TIDPUNKT", "treatment",
            "event_time", "new_event_time", "post", "D"
        ]
    ]
    .sort_values(["cohort", "aID", "new_event_time"])
    .head(100)
)

# ============================================================
# Basic DID
# ============================================================

model_high, result_high = run_basic_did_one(
    matched_high,
    outcome_col=outcome_col,
    cluster_col="aID"
)

print_result_table(result_high, title="Overall DID: High price period")

cohort_basic_models_high, cohort_basic_results_df_high = run_basic_did_by_cohort(
    df=matched_high,
    outcome_col=outcome_col,
    min_treated=30,
    cluster_col="aID"
)

print_result_table(
    cohort_basic_results_df_high,
    title="Cohort DID: High price period"
)

# ============================================================
# Dynamic DID
# ============================================================

event_model_high, event_results_high = run_event_study_one(
    matched_high,
    outcome_col=outcome_col,
    event_window=EVENT_WINDOW,
    reference=REFERENCE_EVENT_TIME,
    event_col="new_event_time",
    cluster_col="aID"
)

print_result_table(
    event_results_high,
    title="Dynamic DID: High price period"
)

cohort_event_models_high, cohort_event_results_df_high = run_event_study_by_cohort(
    df=matched_high,
    outcome_col=outcome_col,
    event_window=(-12, 6),
    reference=-1,
    event_col="new_event_time",
    min_treated=30,
    cluster_col="aID"
)

print_result_table(
    cohort_event_results_df_high,
    title="Dynamic DID by cohort: High price period"
)

plot_event_study(
    event_results_high,
    outcome_col=outcome_col,
    title="Dynamic Treatment Effect",
    ylim=HIGH_YLIM
)

plot_event_study_by_cohort(
    cohort_event_results_df_high,
    outcome_col=outcome_col,
    ylim=HIGH_YLIM
)

# ============================================================
# Dynamic DID by cohort group
# ============================================================

matched_high = assign_cohort_group(matched_high)

group_event_models_high, group_event_results_df_high = run_event_study_by_cohort_group(
    df=matched_high,
    outcome_col=outcome_col,
    event_window=(-12, 6),
    reference=-1,
    event_col="new_event_time",
    min_treated=30,
    cluster_col="aID"
)

print_result_table(
    group_event_results_df_high,
)

plot_event_study_by_cohort_group(
    group_event_results_df_high,
    outcome_col=outcome_col,
    ylim=HIGH_YLIM
)
```

```python
def save_results(
    overall_df,
    cohort_df,
    event_df,
    cohort_event_df,
    group_event_df=None,
    save_path="results",
):
    """
    Save DID result tables to CSV.
    """

    import os

    os.makedirs(save_path, exist_ok=True)

    overall_df.to_csv(
        os.path.join(save_path, "overall_effect.csv"),
        index=False,
    )

    cohort_df.to_csv(
        os.path.join(save_path, "cohort_effect.csv"),
        index=False,
    )

    event_df.to_csv(
        os.path.join(save_path, "dynamic_effect.csv"),
        index=False,
    )

    cohort_event_df.to_csv(
        os.path.join(save_path, "dynamic_cohort_effect.csv"),
        index=False,
    )

    if group_event_df is not None:
        group_event_df.to_csv(
            os.path.join(save_path, "dynamic_cohort_group_effect.csv"),
            index=False,
        )

    print(f"✅ CSV results saved to: {save_path}")
     
```

```python
save_results(
    overall_df=result_high,
    cohort_df=cohort_basic_results_df_high,
    event_df=event_results_high,
    cohort_event_df=cohort_event_results_df_high,
    group_event_df=group_event_results_df_high,
    save_path=f"/lakehouse/default/Files/output/did/{match_method}/results_high"
)
```
